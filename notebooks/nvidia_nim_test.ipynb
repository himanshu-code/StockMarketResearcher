```python
from openai import OpenAI

client = OpenAI(
  base_url = "https://integrate.api.nvidia.com/v1",
  api_key = ""
)

completion = client.chat.completions.create(
  model="openai/gpt-oss-20b",
  messages=[{"role":"user","content":"what is an llm"}],
  temperature=1,
  top_p=1,
  max_tokens=4096,
  stream=False
)
print(completion)
```

```python
from crewai import LLM
import os

# nvidia_nim/ prefix tells LiteLLM to send the full openai/gpt-oss-20b model ID
# to NVIDIA NIM intact. Using openai/ prefix alone causes LiteLLM to strip it -> 404
llm = LLM(
    model='nvidia_nim/openai/gpt-oss-20b',
    api_key="",
    temperature=0
)
```

```python
llm.call("tel me joke related to spiderman")
```
